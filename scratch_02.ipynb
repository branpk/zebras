```python
from PIL import Image
from pathlib import Path

for path in sorted(Path("public/zebras/light").glob("*.jpg")):
    if path.name == "DmXfJ9lt.jpg":
        continue

    image = Image.open(path)
    dark_image = Image.open(Path("public/zebras/dark-v2") / path.name)

    scale = 640 / max(image.width, image.height)
    if scale < 1:
        size = (round(image.width * scale), round(image.height * scale))
        image = image.resize(size)
        dark_image = dark_image.resize(size)

    output_path = Path("public/zebras/light-small") / path.name
    output_path.parent.mkdir(parents=True, exist_ok=True)
    image.save(output_path)

    output_path = Path("public/zebras/dark-v2-small") / path.name
    output_path.parent.mkdir(parents=True, exist_ok=True)
    dark_image.save(output_path)
```
